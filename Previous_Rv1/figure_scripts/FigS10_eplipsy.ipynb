```python
import numpy as np

# ===========================
# 1) Merge CA1 int + CA1 pyr
# ===========================
def get_ca1_population_spikes(spike_times, ca1int_idx, ca1pyr_idx):
    """
    spike_times : list of 1D arrays (all units)
    ca1int_idx  : 1D array of Python indices for CA1 interneurons
    ca1pyr_idx  : 1D array of Python indices for CA1 pyramidal cells

    Returns:
        ca1_units : list of arrays (CA1 cells only)
        ca1_all_spikes : 1D numpy array of all CA1 spikes pooled
    """
    all_idx = sorted(set(ca1int_idx.tolist()) | set(ca1pyr_idx.tolist()))
    ca1_units = [np.asarray(spike_times[i], float).ravel() for i in all_idx]
    # flatten
    ca1_all_spikes = np.concatenate(
        [u[np.isfinite(u)] for u in ca1_units if u.size > 0]
    ) if ca1_units else np.array([], float)
    ca1_all_spikes = np.sort(ca1_all_spikes)
    return ca1_units, ca1_all_spikes

# ---- data (../csv) ----
ca1_units = ca1_all_spikes = None  # spike times are not shipped
# ---- the same from the recordings, with analysis_scripts/: ----
# ca1_units, ca1_all_spikes = get_ca1_population_spikes(spike_times, ca1int_py, ca1pyr_py)
# print(f"CA1 population: {len(ca1_units)} cells, {len(ca1_all_spikes)} spikes.")


# ==================================
# 2) Bin and get population rate
# ==================================
def population_counts(spike_times, bin_size=0.01):
    """
    spike_times : 1D array of pooled spikes (seconds)
    bin_size    : bin width (s) for population activity

    Returns:
        bin_edges, counts, rates
    """
    if spike_times.size == 0:
        return np.array([0.0, bin_size]), np.array([0], int), np.array([0.0], float)

    t_max = float(np.nanmax(spike_times))
    nbins = int(np.ceil(t_max / bin_size))
    edges = np.linspace(0.0, nbins * bin_size, nbins + 1)
    counts, _ = np.histogram(spike_times, bins=edges)
    rates = counts.astype(float) / bin_size  # spikes/s (population)
    return edges, counts, rates

BIN = 0.01  # 40 ms bins for branching-ratio / avalanche style
# ---- data (../csv) ----
edges = pop_counts = pop_rates = None  # spike times are not shipped
# ---- the same from the recordings, with analysis_scripts/: ----
# edges, pop_counts, pop_rates = population_counts(ca1_all_spikes, bin_size=BIN)


# =====================================
# 3) Detect "epileptic" high-rate epochs
# =====================================
def detect_high_rate_epochs(rates, bin_size, z_thresh=3.0, min_dur=1.0):
    """
    Simple heuristic: epileptic epochs = contiguous bins where
    z-scored population rate > z_thresh.

    Returns:
        list of (start_time, end_time)
    """
    r = np.asarray(rates, float)
    mu = np.nanmean(r)
    sd = np.nanstd(r, ddof=1)
    if sd == 0 or not np.isfinite(sd):
        return []

    z = (r - mu) / sd
    above = z > z_thresh

    epochs = []
    n_min = int(np.ceil(min_dur / bin_size))
    i = 0
    while i < len(above):
        if above[i]:
            j = i + 1
            while j < len(above) and above[j]:
                j += 1
            if j - i >= n_min:
                start_t = i * bin_size
                end_t   = j * bin_size
                epochs.append((start_t, end_t))
            i = j
        else:
            i += 1

    return epochs

epi_epochs=[(108*60, 118*60)] #manual definition of epileptic epochs
#print("Epileptic candidate epochs:", epi_epochs)
# =========================================
# 4) Branching ratio estimator
# =========================================
def estimate_branching_ratio(counts):
    """
    Simple branching-ratio estimator:
    sigma = sum(n_{t+1}) / sum(n_t) over bins with n_t > 0.
    """
    c = np.asarray(counts, float)
    if c.size < 2:
        return np.nan
    parent = c[:-1]
    child  = c[1:]
    mask = parent > 0
    if not np.any(mask):
        return np.nan
    sigma = child[mask].sum() / parent[mask].sum()
    return float(sigma)

def branching_sliding(edges, counts, win=10.0, step=2.0, interval=None):
    """
    Compute branching ratio in sliding windows.

    Parameters
    ----------
    edges : array
        Bin edges for counts (len = nbins + 1), in seconds.
    counts : array
        Population spike counts per bin (len = nbins).
    win : float
        Window length in seconds (default 10.0).
    step : float
        Step between window starts in seconds (default 2.0).
    interval : (start, end) or None
        If given, restrict sliding windows to this time interval (seconds).
        If None, use the full recording.

    Returns
    -------
    results : list of dict
        Each dict has keys: 'start', 'end', 'sigma'
    """
    counts = np.asarray(counts, float)
    centers = 0.5 * (edges[:-1] + edges[1:])

    # define overall time range
    if interval is None:
        t_start = float(centers[0])
        t_end   = float(centers[-1])
    else:
        t_start, t_end = map(float, interval)

    results = []
    t = t_start
    while t + win <= t_end:
        a = t
        b = t + win
        mask = (centers >= a) & (centers < b)
        if mask.sum() >= 2:
            sigma = estimate_branching_ratio(counts[mask])
        else:
            sigma = np.nan

        results.append({"start": a, "end": b, "sigma": sigma})
        t += step

    return results

# branching ratio inside each high-rate (putative epileptic) epoch
epi = epi_epochs[0]  # e.g. (102.0, 145.0)
# ---- data (../csv) ----
sliding_epi = None  # (not used by any panel)
# ---- the same from the recordings, with analysis_scripts/: ----
# sliding_epi = branching_sliding(edges, pop_counts, win=20.0, step=2.0, interval=epi)


# =========================================
# 5) Random control intervals outside epilepsy
# =========================================
import random

def sample_control_epochs(total_dur, bad_epochs, length, n_samples=20, rng=None):
    """
    Randomly sample non-overlapping control epochs of given length that
    do NOT overlap any bad_epochs.
    total_dur : recording duration (s)
    bad_epochs: list of (start, end) to avoid
    length    : length of each control epoch (s)
    """
    rng = random.Random(rng)
    controls = []
    # helper to test overlap
    def overlaps(a, b):
        return not (a[1] <= b[0] or a[0] >= b[1])

    attempts = 0
    while len(controls) < n_samples and attempts < 1000:
        start = rng.uniform(0, total_dur - length)
        cand = (start, start + length)
        if any(overlaps(cand, ep) for ep in bad_epochs):
            attempts += 1
            continue
        controls.append(cand)
    return controls

# ---- data (../csv) ----
total_dur = None  # (not used by any panel)
# ---- the same from the recordings, with analysis_scripts/: ----
# total_dur = edges[-1]
# choose control length = median epileptic epoch length (or 2 s if none)
if epi_epochs:
    epi_lengths = [e[1] - e[0] for e in epi_epochs]
    ctrl_len = float(np.median(epi_lengths))
else:
    ctrl_len = 2.0

control_epochs = [(14*60,24*60)]
```

```python
import numpy as np
import pandas as pd
import os
from criticality import WiltingPriesemann as WP
import pickle
import tempfile
import time
def compute_and_plot_branching_ratios(interval_matrices, time_intervals, k_max=100,
                                      output_dir="branching_ratio_plots"):
    """
    Computes and plots branching ratios for each subinterval.

    Parameters:
    - interval_matrices: list of arrays
        Each element is either:
          * 2D spike matrix: shape (n_neurons, T), or
          * 1D activity vector A_t: shape (T,)
    - time_intervals: list of (start, end) tuples for each subinterval.
    - k_max: Maximum time lag for branching ratio calculation.
    - output_dir: Directory to save plots.

    Returns:
    - results: dict with keys:
        'centers'           : list of window centers
        'branching_ratios'  : list of branching ratios (float or None)
        'Goodfit'           : list of fit metrics (one per interval)
    """
    branching_ratios = []
    results = []
    centers = []
    Goodfit = []

    os.makedirs(output_dir, exist_ok=True)

    for i, (matrix, interval) in enumerate(zip(interval_matrices, time_intervals)):
        matrix = np.asarray(matrix)

        # ----- Get A_t (activity per bin) -----
        if matrix.ndim == 1:
            # already a population activity vector
            A_t = matrix
        else:
            # 2D spike matrix: count active neurons per bin
            A_t = np.count_nonzero(matrix, axis=0)

        # Skip intervals with insufficient activity
        if len(A_t) < 2:
            branching_ratios.append(None)
            centers.append((interval[0] + interval[1]) / 2)
            Goodfit.append(None)
            continue

        # Use WP.MR_estimation to compute branching ratio
        mr_result = WP.MR_estimation(A_t, k_max)

        # Compute the center of the interval
        center = (interval[0] + interval[1]) / 2.0

        branching_ratio = mr_result['branching_ratio']
        k = mr_result['k']
        rk_data = mr_result['r_k']
        b = mr_result['p_opt'][0]  # Fitted b
        m = mr_result['p_opt'][1]  # Fitted m


        # Save the results
        centers.append(center)
        branching_ratios.append(branching_ratio)

    results = {
        "centers": centers,
        "branching_ratios": branching_ratios,
        "Goodfit": Goodfit,
    }
    return results
def make_subinterval_counts(edges, counts, interval, win=20.0, step=2.0):
    """
    Slice global population counts into sliding windows inside a given interval.

    Parameters
    ----------
    edges : array, shape (nbins+1,)
        Bin edges in seconds.
    counts : array, shape (nbins,)
        Population spike counts per bin.
    interval : (start, end)
        Time interval in seconds to restrict windows to.
    win : float
        Window length in seconds.
    step : float
        Step size between window starts in seconds.

    Returns
    -------
    interval_counts : list of 1D arrays
        Each is counts for one window.
    time_intervals  : list of (start, end) tuples for each window.
    """
    counts = np.asarray(counts, float)
    centers = 0.5 * (edges[:-1] + edges[1:])

    t_start, t_end = map(float, interval)

    interval_counts = []
    time_intervals = []

    t = t_start
    while t + win <= t_end:
        a = t
        b = t + win
        mask = (centers >= a) & (centers < b)
        if mask.sum() >= 2:
            interval_counts.append(counts[mask].copy())
            time_intervals.append((a, b))
        t += step

    return interval_counts, time_intervals
WIN  = 20.0
STEP = 2.0

# ---- data (../csv) ----
# branching ratio (MR estimator; CA1 population counts in 10 ms bins, k_max = 100, i.e. lags up to 1 s)
# of every 20 s window, stepped 2 s, inside the epileptic and the control epoch
_br = read_csv('branching_ratio_windows.csv')
epi_intervals, ctl_intervals = (list(zip(_br.loc[_br['epoch'] == e, 'start'], _br.loc[_br['epoch'] == e, 'end'])) for e in ('epileptic', 'control'))
epi_results, ctl_results = ({'centers': _br.loc[_br['epoch'] == e, 'center'].tolist(),
                             'branching_ratios': _br.loc[_br['epoch'] == e, 'branching_ratio'].tolist(),
                             'Goodfit': []} for e in ('epileptic', 'control'))
# ---- the same from the recordings, with analysis_scripts/: ----
# # Build sliding-window count segments for all epochs
# epi_counts_list, epi_intervals = [], []
# for epi in epi_epochs:
#     counts, intervals = make_subinterval_counts(
#         edges, pop_counts, epi, win=WIN, step=STEP
#     )
#     epi_counts_list.extend(counts)
#     epi_intervals.extend(intervals)
# 
# ctl_counts_list, ctl_intervals = [], []
# for ctl in control_epochs:
#     counts, intervals = make_subinterval_counts(
#         edges, pop_counts, ctl, win=WIN, step=STEP
#     )
#     ctl_counts_list.extend(counts)
#     ctl_intervals.extend(intervals)
# 
# print(f"EPI windows:  {len(epi_counts_list)}")
# print(f"CTRL windows: {len(ctl_counts_list)}")
# 
# epi_results = compute_and_plot_branching_ratios(
#     epi_counts_list,
#     epi_intervals,
#     k_max=100,
#     output_dir="branching_ratio_plots_epi"
# )
# 
# ctl_results = compute_and_plot_branching_ratios(
#     ctl_counts_list,
#     ctl_intervals,
#     k_max=100,
#     output_dir="branching_ratio_plots_ctrl"
# )
```
