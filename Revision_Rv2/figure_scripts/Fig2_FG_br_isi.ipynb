```python
# Data: ../csv/Fig2_FG_br_isi/
from pathlib import Path
import pandas as pd
CSV = Path('..') / 'csv' / 'Fig2_FG_br_isi'


def read_csv(name, **kw):
    """A table of this notebook (round_trip: the floats exactly as written)."""
    return pd.read_csv(CSV / name, float_precision='round_trip', **kw)

import os, sys
sys.path.insert(0, os.path.abspath('.'))   # figure_scripts/
import final_figs as F                     # print styles, saving, stars
```

## Fig 2F — branching ratio around SWRs, ripple-excised

Peri-SWR branching ratio (MR estimator, 4 ms bins, active pyramidal units, k_max 200) on 10 s windows stepped 2 s, each window's value from its ripple-free sub-windows; example PPP21 day 5, SWRs vs shuffle, ± SEM over ripples; inset: ΔBR per session, tested on per-session means.

- Drawing: `fig2_new.panel_psth`; data `fig2g_psth_curves_PPP21_day5.csv`, `fig2g_psth_nine_sessions.csv` (producer `fig2g_psth_excised.py`).

```python
import pandas as pd
import matplotlib.pyplot as plt
import fig2_new_drawing as N            # fig2_new.py's drawing functions (copied)
N.LOGDIR = str(CSV)                        # they read their tables from this notebook's csv folder

# Each panel: the producer's own panel function, drawn under its native settings at the final figsize, then
# restyled for print by F.shrink (fonts 11/10/9 pt -> 6/5.5/5, lines x 0.5); only labels too long for the
# printed size are shortened.

# ---- F: branching ratio around SWRs, ripple-excised
F.native_style()
fig, ax = plt.subplots(figsize=(46 * F.MM, 48 * F.MM), layout="constrained")
N.panel_psth(ax)
ax.set_ylabel("Branching ratio\n(ripple-excised)")  # shortened; this is the example-session trace
for t in [t for t in ax.texts if t.get_text().startswith("example ")]:
    t.remove()                                       # "example PPP21_day5; shade = SEM over ripples": wider than the axes at print size -> caption
ax.legend(frameon=False, loc="lower left")           # positioned off the curve at print size
F.shrink(fig)
F.sparse_yticks(ax)                          # every other y label
F.settle(fig)
F.drop_counts(fig)                          # formal panel: no k/n; counts in the legend
F.show(F.save(fig, "Fig2_F_br_around_swr_bgd2"))
```

## Fig 2G — ISI entropy by SWR-rate sextile, full window and ripple-excised

ISI entropy (bits) of the pooled CA1 spike train per non-overlapping 30 s NREM window, by within-session SWR-rate sextile; full window (grey) and ripple-excised (red); 9 decoded sessions; per-session Spearman, signed-rank test.

- Data `fig2_agh_nine_sessions.csv` (producer `fig2_agh_excised.py`); drawing `fig2_composite.sextile_panel`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import fig2_new_drawing as N            # fig2_new.py's drawing functions (copied)
N.LOGDIR = str(CSV)                        # they read their tables from this notebook's csv folder

# ---- G: ISI entropy by SWR-rate sextile, full window and ripple-excised
S = pd.read_csv(os.path.join(N.LOGDIR, "fig2_agh_nine_sessions.csv"))
F.native_style()
fig, ax = plt.subplots(figsize=(40 * F.MM, 48 * F.MM), layout="constrained")
N.sextile_panel(ax, S, "isi_entropy_pooled", "ISI entropy (bits)", arms=(("full window", N.GREY),), corner="right")
N.sextile_panel(ax, S, "isi_excised", "ISI entropy (bits)", arms=(("ripple-excised", N.RED),), corner="right")
# the statistic for each arm moves into that arm's legend label, drawn in the order of the arms (drop_counts then
# produces the formal legend with just the significance stars)
stats = list(ax.texts)
handles, labels = ax.get_legend_handles_labels()
assert len(stats) == len(handles) == 2, (len(stats), len(handles))
for h, lab, t in zip(handles, labels, stats):
    h.set_label(f"{lab}: {t.get_text()}")
    t.remove()
ax.legend(frameon=False, loc="lower left")
F.shrink(fig)
for t in ax.get_legend().get_texts():               # coloured to match each arm's line
    t.set_color(N.RED if t.get_text().startswith("ripple") else N.GREY)
F.settle(fig)
F.drop_counts(fig)                          # formal panel: no k/n; counts in the legend
F.show(F.save(fig, "Fig2_G_isi_entropy"))
```
