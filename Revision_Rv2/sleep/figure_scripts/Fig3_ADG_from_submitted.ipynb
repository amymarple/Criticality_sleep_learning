# Figure 3 a–d and the g example

Panels a–d are the network-model row of Weiwei Chen (`assets/Fig3model_Weiwei_2026-10-17.svg`), 174.7 × 39.3 mm at
final size; the left half of g is cut from Figure 3 of the first submission (`assets/Figure3_submitted_2026-03-12.svg`).
Each region becomes a panel file at the figure's own size, without its panel letter (`F.crop_svg`).

- g example (CCK⁺ silencing schematic, CCK⁺ / CA1 pyr raster, 100 ms bar): units x 349–453.5, y 160–300: 27.6 × 37.0 mm.
- The model figures behind a–d are drawn in [`../../behavior`](../../behavior) (Figure 3c, d; Extended Data Fig. 8).

```python
# Data: ../csv/Fig3_ADG_from_submitted/
from pathlib import Path
import pandas as pd
CSV = Path('..') / 'csv' / 'Fig3_ADG_from_submitted'


def read_csv(name, **kw):
    """A table of this notebook (round_trip: the floats exactly as written)."""
    return pd.read_csv(CSV / name, float_precision='round_trip', **kw)

import os, sys
sys.path.insert(0, os.path.abspath('.'))   # figure_scripts/
import final_figs as F                     # print styles, saving, stars
```

```python
# a-d: the network-model row of Weiwei Chen, an Illustrator export cropped to the top row (1 unit = 1 pt, so
# 174.7 x 39.3 mm at final size). Its letters (no ids) are left out, and its CSS classes are renamed so that its
# stylesheet stays inside the panel. b's colour bar: the export writes its 12 x 216 px gradient as 11 x 218 at
# scale(.3), i.e. 3.3 x 65.4 units, short of the bar it is clipped to (x 234.3, y 18.4, 3.6 x 71.2), so the gradient
# is placed on that clip rectangle.
# g example: cut from the submitted Figure 3 (Illustrator's "Figure 3.pdf" of 2026-03-12, imported by Inkscape with
# its text kept as text), in its own units (1 unit = 1/96 in = 0.265 mm); its letter is left out.
import re

# ---- data (../csv) ----
MODEL = str(CSV / "assets" / "Fig3model_Weiwei_2026-10-17.svg")   # the model row
SRC = str(CSV / "assets" / "Figure3_submitted_2026-03-12.svg")     # the submitted Figure 3 (Inkscape import of its PDF)
# ---- the same from the recordings, with analysis_scripts/: ----
# MODEL = os.path.join(F.FINAL, "Fig3", "source", "Fig3model_Weiwei_2026-10-17.svg")
# SRC_PDF = os.path.join(F.FINAL, "Fig3", "source", "Figure3_submitted_2026-03-12.pdf")
# SRC = SRC_PDF[:-4] + ".svg"
# if not os.path.exists(SRC):
#     import subprocess
#     subprocess.run([F.INKSCAPE, SRC_PDF, "--export-type=svg", "--export-plain-svg", f"--export-filename={SRC}"],
#                    check=True, capture_output=True)      # default (internal) PDF import: text stays text
G_EX = (349, 160, 104.5, 140)          # g's CCK example (title, schematic, raster, 100 ms bar): ink x 349-452.7,
                                       # y 164.7-297.7; the box plot's y label starts at x 454.7
ms = open(MODEL, encoding="utf-8").read()
MODEL_LETTERS = re.findall(r'<text class="cls-26"[^>]*>(?:(?!</text>).)*</text>', ms, flags=re.S)
assert sorted(re.sub(r"<[^>]+>", "", t) for t in MODEL_LETTERS) == ["a", "b", "c", "d"]
AD = tuple(float(v) for v in re.search(r'viewBox="([^"]+)"', ms).group(1).split())     # (0, 0, 495.1, 111.4)
F.crop_svg(MODEL, "Fig3_AD_model_and_inputs", *AD, unit_mm=25.4 / 72,
           edits=[(t, "") for t in MODEL_LETTERS] + [("cls-", "f3m-"),
                  ('<image width="11" height="218" transform="translate(234.3 18.4) scale(.3)"',
                   '<image preserveAspectRatio="none" x="234.3" y="18.4" width="3.6" height="71.2"')])
F.crop_svg(SRC, "Fig3_G_cck_silencing_example", *G_EX,
           drop_ids=["tspan5784", "text3188", "text3192", "text3196"])   # its letter "G"; the box plot's "d2 (z)" label
F.show(F.stem("Fig3_AD_model_and_inputs") + ".png")
F.show(F.stem("Fig3_G_cck_silencing_example") + ".png")
```
