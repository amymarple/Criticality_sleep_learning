# Extended Data Fig. 4: d2 and the branching ratio

Reads its tables from `../csv/FigS4_d2_branching_ratio/`. Panels only; saved to `../figure_output/supplementary/FigS4/`. d2 = d2 (2024 Euclidean, AR p = 20, 40 ms), labelled "d2"; m = branching ratio, MR estimator, 4 ms bins, k_max = 200 (800 ms lag range). Non-overlapping 30 s windows, NREM except in **b**; session is the unit. Each figure is drawn by its producer's drawing code and cut into single panels (`si_panels.panel`). Stars: *p < 0.05, **p < 0.01, ***p < 0.001, ****p < 0.0001.

- **a** m (all CA1 units, the population d2 is fitted on) against d2 on the same windows, one colour and one least-squares line per session, 18 sessions (`figs4a_scatter_windows.csv`).
- **b** Per-session Pearson r(d2, m) within each state, m as in **a** (NREM 18, REM 12, Wake θ 7, Wake non-θ 18 sessions; a wake window lies wholly in θ or non-θ); Fisher-z mean ± s.e.m. (`fig1_d2_vs_br.csv`).
- **c, d** d2 (**c**) and m on pyramidal cells (**d**) by within-session sextile of SWR rate, full window (grey) and ripple-excised (red), mean ± s.e.m. over 16 sessions; per-session Spearman ρ with SWR rate, two-sided Wilcoxon signed-rank test across sessions (`d2_arms_raw_windows.csv`, `r24_br_ripple_windows.csv`).
- **e–g** d2 regressed on m (pyramidal) within each session; d2 and the residual d2 | m correlated with **e** the replay-SWR rate given SWR rate (partial Spearman ρ; 9 decoded sessions; `excised_br_entropy_windows.csv`), **f** SWR rate (16 sessions) and **g** the barrage rate given SWR rate (12 sessions; excised = ripples and barrages, blue; `r16_event_excised_windows.csv`, `r16_event_excised_windows_br.csv`), full window (grey) and events excised; Fisher-z mean ± s.e.m.; two-sided Wilcoxon signed-rank test of the per-session ρ against 0 (n.s. where p ≥ 0.05).

```python
# Data: ../csv/FigS4_d2_branching_ratio/
from pathlib import Path
import pandas as pd
CSV = Path('..') / 'csv' / 'FigS4_d2_branching_ratio'


def read_csv(name, **kw):
    """A table of this notebook (round_trip: the floats exactly as written)."""
    return pd.read_csv(CSV / name, float_precision='round_trip', **kw)

import os, sys
sys.path.insert(0, os.path.abspath('.'))   # figure_scripts/
import matplotlib.pyplot as plt
import final_figs as F                     # print styles, saving, stars
from si_panels import producer, grab, panel   # a drawing module's figure -> single panels at print size

plt.rcParams.update({"svg.fonttype": "none", "pdf.fonttype": 42})   # text stays text
```

## a–d: m against d2, and both against SWR rate

```python
for k in ("FIG_ALL", "POP", "TOP", "CORR_POP", "ALL", "BINS"):
    os.environ.pop(k, None)                              # every drawing module at its defaults

# a-d: r24_figure.py's figure (m on all CA1 units in a, b; pyramidal m in d), drawn by its drawing code, stars by F.stars
B = producer("r24_br_ripple_arms_drawing")             # r24_br_ripple_arms.py: SWR-rate panels and their statistics
M = producer("r24_figure_drawing")                     # r24_figure.py: m vs d2, r(d2, m) by state
B.LOGDIR = M.LOGDIR = str(CSV)                           # they read their tables from this notebook's csv folder
B.stars = F.stars
# ---- data (../csv) ----
# r24_figure.main(): the Fig 1 NREM windows with d2 and m (a), fig1_d2_vs_br.csv (b), d2_arms_raw_windows.csv and
# r24_br_ripple_windows.csv (c, d)
FIG_AD = grab(lambda: M.draw_main(read_csv("figs4a_scatter_windows.csv")), "r24_d2_vs_br")
# ---- the same from the recordings, with analysis_scripts/: ----
# B = producer("r24_br_ripple_arms"); M = producer("r24_figure"); B.stars = F.stars
# FIG_AD = grab(M.main, "r24_d2_vs_br")
panel(FIG_AD, 0, "FigS4_A_br_vs_d2", 86, 58, xlabel="d2", ylabel="Branching ratio (m)")
panel(FIG_AD, 1, "FigS4_B_r_d2_br_by_state", 86, 58, ylabel="Pearson r (d2, m)")
panel(FIG_AD, 2, "FigS4_C_d2_by_swr_rate", 86, 52, xlabel="SWR rate sextile", ylabel="d2")
panel(FIG_AD, 3, "FigS4_D_br_by_swr_rate", 86, 52, xlabel="SWR rate sextile", ylabel="Branching ratio (m)")
plt.close(FIG_AD)
```

## e–g: residual d2 | m against replay-SWR, SWR and barrage rate

```python
# e-g: r24_residual_test.py's figure (pyramidal m), barrage column blue, drawn by its drawing code, stars by F.stars
M = producer("r24_residual_test_drawing")
M.LOGDIR = str(CSV)                                      # its cohorts() reads its tables from this notebook's csv folder
M.stars = F.stars
# ---- data (../csv) ----
# r24_residual_test.main(): excised_br_entropy_windows.csv (e), d2_arms_raw_windows.csv and r24_br_ripple_windows.csv (f),
# r16_event_excised_windows.csv and r16_event_excised_windows_br.csv (g)
FIG_EG = grab(M.draw_main, "r24_residual_test")
# ---- the same from the recordings, with analysis_scripts/: ----
# M = producer("r24_residual_test"); M.stars = F.stars
# FIG_EG = grab(M.main, "r24_residual_test")
panel(FIG_EG, 0, "FigS4_E_residual_replay_rate", 57, 52, title="Replay-SWR rate | SWR rate", ylabel="Spearman ρ")
panel(FIG_EG, 1, "FigS4_F_residual_swr_rate", 57, 52, title="SWR rate", ylabel="Spearman ρ")
panel(FIG_EG, 2, "FigS4_G_residual_barr_rate", 57, 52, title="BARR rate | SWR rate", ylabel="Spearman ρ")
plt.close(FIG_EG)
```
